```python
import pandas as pd
import numpy as np
df = pd.DataFrame({ 'Age': [25, 30, np.nan, 40, 35],
     'Department': ['HR', 'Finance', 'Finance', np.nan, 'IT' ]})
print("Original Dataset (with Missing Value):")
print(df)
df['Age']=df['Age'].fillna(df['Age'].mean())
df['Department']=df['Department'].fillna(df['Department'].mode()[0])
print(df)
```

```python
import pandas as pd
import numpy as np
df = pd.DataFrame({ 'Age': [25, 30, np.nan, 40, 35],
     'Department': ['HR', 'Finance', 'Finance', np.nan, 'IT' ]})
print("Original Dataset (with Missing Value):")
print(df)
df_ffill = df.copy()
df_ffill.ffill(inplace=True)
print(df_ffill)
df_bfill = df.copy()
df_bfill.bfill(inplace=True)
print(df_bfill)
```

```python
import pandas as pd
import numpy as np
df = pd.DataFrame({ 'Age': [25, 30, np.nan, 40, 35],
     'Department': ['HR', 'Finance', 'Finance', np.nan, 'IT' ]})
print("Original Dataset")
print(df)
df_drop_rows = df.dropna()
print("After dropping rows:\n", df_drop_rows)
df_drop_cols = df.dropna()
print("After dropping cols:\n", df_drop_cols)
```

```python
import pandas as pd
df = pd.DataFrame({
    'ID': [1, 2, 2, 3, 4, 4],
    'Name': ['Alice', 'Bob', 'Bob', 'Charlie', 'David', 'David'],
    'Age': [25, 30, 30, 35, 40, 40]
})
print("Original Data:\n", df)
df_exact = df.drop_duplicates()
print("\nAfter Exact Match Removal:\n", df_exact)
```

```python
import pandas as pd
df = pd.DataFrame({
    'ID': [1, 2, 2, 3, 4, 4],
    'Name': ['Alice', 'Bob', 'Bob', 'Charlie', 'David', 'David'],
    'Age': [25, 30, 30, 35, 40, 40]
})
print("Original Data:\n", df)
df_subset_id = df.drop_duplicates(subset=['ID'])
print("\nAfter Subset-Based Removal:\n", df_subset_id)
df_subset_name = df.drop_duplicates(subset=['Name'])
print("\nAfter Subset-Based Removal:\n", df_subset_name)
```

```python
import pandas as pd
df = pd.DataFrame({
    'Data': ['2025-01-05', '05/01/2025', 'Jan 5,2025', '2025.01.05']})
print("Original Data:\n", df)
df['Data'] = pd.to_datetime(df['Data'], errors='coerce').dt.strftime('%Y-%m-%d')
print(df)
```
